```python
import pandas as pd
```

```python
data = pd.read_csv("weight-height.csv")
```

```python
data.head()
```

```python
data = data[['Height','Weight']]
```

```python
data
```

```python
import matplotlib.pyplot as plt
```

```python
X = data.Height
y = data.Weight
```

```python
plt.scatter(X,y)
```

```python
def line(x, m=0, b=0):
    pred_y = m*x + b
    return pred_y
```

```python
py = line(X)
```

```python
plt.scatter(X,y)
plt.scatter(X,py)
```
